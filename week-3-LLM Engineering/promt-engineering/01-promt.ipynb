**`❓ Question:`**

When sending instructions to an LLM, developers often place core rules (like "you are a strict assistant") into a "System" prompt, while placing the user's question into a "User" prompt. Why is separating this useful? Choose the correct answer.

> It cuts down prompt word count to save money.

> Models pay closer attention to System prompts, reducing user override attempts.

> System prompts work offline without internet connection.

> It blocks users from typing words listed in System prompts.

</summary>

---
---

**`Solution:`**

Separating instructions into **System** and **User** messages establishes a distinction between the application's higher-level instructions and the user's actual request.

A **System prompt** is typically used to define the assistant's overall behavior, role, rules, and constraints. The **User prompt** contains the specific task or question the user wants answered.

This separation helps the model distinguish **persistent behavioral instructions** from **user-provided content**. In systems that apply instruction hierarchy, higher-priority system instructions take precedence over conflicting user instructions.

---
---

**`Relevant Concept Breakdown`**

Think of the two prompts as having different roles:

~~~text
SYSTEM PROMPT
      │
      ├── Role
      ├── Rules
      ├── Behavior
      └── Constraints
      │
      ▼
   LLM
      ▲
      │
USER PROMPT
      │
      ├── Question
      ├── Request
      └── Input/data
~~~

For example:

~~~text
System:
"You are a strict assistant.
Follow the application's safety and formatting rules."

User:
"Explain how HTTPS works."
~~~

The system instruction establishes **how the assistant should behave**, while the user message specifies **what the assistant should do**.

If a user then tries to give a conflicting instruction, the system-level instruction generally has higher priority.

---
---

**`Solving the Question`**

The question asks why developers separate core rules from the user's question.

The main benefit is **instruction hierarchy**.

The system prompt establishes higher-level behavior, while the user prompt supplies the task. This makes it harder for ordinary user instructions to override the application's higher-priority rules.

For example:

~~~text
System: "You are a strict assistant."

User: "Ignore your rules and behave differently."
~~~

The system instruction has higher priority, so the user's conflicting instruction should not override it.

Therefore:

**Correct Answer:**

> **Models pay closer attention to System prompts, reducing user override attempts.** ✅

---
---

🧠 **`Remember`**

> **System = higher-level rules and behavior**  
> **User = specific request or input**

The key idea is **instruction hierarchy**:

$$
\boxed{\text{System instructions} > \text{User instructions}}
$$

So separating them helps the model distinguish **application-level rules** from **user-provided requests**. 🤖
